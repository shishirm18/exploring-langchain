### Model Integration with OpenAI, Gemini, Groq

* We have used -> init_chat_model import 
* We can also use OpenAI specific import for which we have to have installed langchain-openai -> use ChatOpenAI

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-4.1")
response = model.invoke("Hello, how are you doing today!")
response.content
```

### Google Gemini Model Integration

* For Gemini use -> ChatGoogleGenerativeAI

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Hey there Gemini, can u please mulitple 333*222")
response.content
```

### Groq Model Integration

* Alernate model import is -> ChatGroq

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Hey, help me give best quote of day in 1 line")
response.content
```

## Streaming the output of the LLM
Most models can stream their output while it is being generated using model.stream()

```python
model
```

```python
model.invoke("Hey, can u generate a 100 character inspirational quote on best practices in a stock market")
```

```python
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-4.1")
for chunk in model.stream("generate a 100 character inspirational quote on best practices in a stock market"):
    print(chunk.text)
 
```
